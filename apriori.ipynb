```python
import numpy as np
import pandas as pd
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("/content/market_basket_dataset.csv")
df.head()
```

```python
# The dataset has a column 'Items' with lists stored as strings, so convert them back
df["Items"] = df["Items"].apply(lambda x: x.strip("[]").replace("'", "").split(", "))

# Convert transactions into a list of lists
transactions = df["Items"].tolist()
```

```python
#Encode the dataset into one-hot format
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)
df_encoded = pd.DataFrame(te_data, columns=te.columns_)
```

```python
# Apply Apriori to find frequent itemsets
# min_support = 0.05 means itemsets appearing in at least 5% of transactions
frequent_itemsets = apriori(df_encoded, min_support=0.05, use_colnames=True)
```

```python
print("Frequent Itemsets:")
print(frequent_itemsets)
```

```python
#Generate association rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)
```

```python
# Show support, confidence, and lift
print("\nAssociation Rules (with support, confidence, and lift):")
print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])
```

```python
# Sorting rules by lift to see strongest associations
rules_sorted = rules.sort_values(by="lift", ascending=False)
print("\nTop Association Rules sorted by Lift:")
print(rules_sorted[["antecedents", "consequents", "support", "confidence", "lift"]].head(10))
```
